```python
import os
import shutil
import random

# Configuration
source_root = 'dataset'
target_root = 'dataset'
classes = ['CocoScale', 'Healty']
train_ratio = 0.8

for cls in classes:
    # 1. Create directories
    os.makedirs(os.path.join(target_root, 'train', cls), exist_ok=True)
    os.makedirs(os.path.join(target_root, 'test', cls), exist_ok=True)
    
    # 2. Get list of all images in the class folder
    src_path = os.path.join(source_root, cls)
    images = [f for f in os.listdir(src_path) if os.path.isfile(os.path.join(src_path, f))]
    
    # 3. Shuffle and Split
    random.shuffle(images)
    split_idx = int(len(images) * train_ratio)
    train_images = images[:split_idx]
    test_images = images[split_idx:]
    
    # 4. Copy files to new destination
    for img in train_images:
        shutil.copy(os.path.join(src_path, img), os.path.join(target_root, 'train', cls, img))
    for img in test_images:
        shutil.copy(os.path.join(src_path, img), os.path.join(target_root, 'test', cls, img))

print(f"Successfully split images! Check the '{target_root}' folder.")
```
